## Task 4

Ask Open-Meteo's historical weather API (`https://archive-api.open-meteo.com/v1/archive`) for the daily maximum temperature in Lisbon from 1 to 7 July 2024. Use latitude 38.72, longitude -9.14, `start_date` and `end_date`, `temperature_2m_max` as the `daily` variable, and `Europe/Lisbon` as the `timezone`.

Pass the six values as a dict to `params`, with `timeout=10`, and save the reply as `r`. Print `r.url` and `r.status_code`. Then print the hottest day of that week, its maximum temperature and the unit, for example `2024-07-01 26.7 °C`. Find the hottest day with code, not by reading the list.

````python
# Task 4
import requests

url = "https://archive-api.open-meteo.com/v1/archive"
params = {
    "latitude": 38.72,
    "longitude": -9.14,
    "start_date": "2024-07-01",
    "end_date": "2024-07-07",
    "daily": "temperature_2m_max",
    "timezone": "Europe/Lisbon",
}

r = requests.get(url, params=params, timeout=10)
print(r.url)
print(r.status_code)

weather = r.json()
print(weather["daily"]["temperature_2m_max"], weather["daily_units"]["temperature_2m_max"])

temps = weather["daily"]["temperature_2m_max"]
i = max(range(len(temps)))
print(f'{weather["daily"]["time"][i]} {temps[i]} '
      f'{weather["daily_units"]["temperature_2m_max"]}')
````
